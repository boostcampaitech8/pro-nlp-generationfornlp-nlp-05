```python
import pandas as pd
from collections import Counter

df_output = pd.read_csv('TestSet_Inference_Qwen3-32B.csv')
df_submission = pd.read_csv('output.csv')

list_choice = []
for i in range(len(df_output)):
    for r in range(10):
        for s in range(20):
            try:
                choice = df_output.loc[i, f'resp_{r}_{s}'].split('{"정답": "')[-1].split('"}')[0]
                if choice in ['1', '2', '3', '4', '5']:
                    list_choice.append(choice)
            except:
                pass

    print()
    list_choice.sort()
    count_choices = Counter(list_choice)
    top1 = count_choices.most_common(2)

    try:
        df_submission.loc[i, 'answer'] = top1[0][0]
    except:
        df_submission.loc[i, 'answer'] = '1'
    
df_submission.to_csv(f"submission.csv", index=False)
```
