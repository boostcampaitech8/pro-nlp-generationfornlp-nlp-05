```python
from FlagEmbedding import FlagReranker

reranker = FlagReranker("dragonkue/bge-reranker-v2-m3-ko", use_fp16=True, devices=['cuda:0'])

def rerank512(query: str, docs: list[str]):
    pairs = [[query, d] for d in docs]
    scores = reranker.compute_score(
        sentence_pairs=pairs,
        batch_size=128,
        max_length=1024,
        normalize=True,
    )
    ranked = sorted(zip(docs, scores), key=lambda x: x[1], reverse=True)
    return ranked
```
